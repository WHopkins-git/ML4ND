# Introduction to Neutron Cross-Section Data

## What are Nuclear Cross-Sections?

Nuclear cross-sections are fundamental quantities in nuclear physics that describe the probability of a nuclear reaction occurring. Think of them as a measure of the "effective area" that a nucleus presents to an incoming particle.

### Key Concepts:

- **Cross-section (σ)**: Measured in barns (1 barn = 10⁻²⁴ cm²)
- **Energy dependence**: Cross-sections vary dramatically with the energy of the incoming neutron
- **Reaction types**: Different reactions have different cross-sections:
  - Total cross-section (σ_tot)
  - Elastic scattering (σ_elastic)
  - Capture (σ_γ)
  - Fission (σ_f)
  - Inelastic scattering (σ_inelastic)

## Nuclear Data Sources

In this tutorial series, we'll explore several major nuclear data sources:

### 1. **ENDF (Evaluated Nuclear Data File)**
- **Type**: Evaluated (validated) data
- **Maintained by**: Cross Sections Evaluation Working Group (CSEWG)
- **Coverage**: Comprehensive nuclear data for various applications
- **Access**: Python packages like `openmc`, `sandy`, `ENDFtk`

### 2. **EXFOR (Experimental Nuclear Reaction Data)**
- **Type**: Experimental (raw) data
- **Maintained by**: IAEA Nuclear Data Section
- **Coverage**: Experimental measurements from publications
- **Access**: IAEA web API, `x4i3` Python package

### 3. **TENDL (TALYS-based Evaluated Nuclear Data Library)**
- **Type**: Evaluated data
- **Maintained by**: PSI/IAEA
- **Coverage**: Extensive isotope coverage
- **Access**: Direct download, similar format to ENDF

### 4. **JANIS (JAva-based Nuclear Data Information System)**
- **Type**: Web interface to multiple databases
- **Maintained by**: OECD NEA
- **Coverage**: Aggregates multiple libraries
- **Access**: Web API

## Why Machine Learning?

Machine learning can assist in the nuclear data validation process by:

1. **Anomaly Detection**: Identifying unusual patterns in experimental data
2. **Interpolation**: Predicting cross-sections at unmeasured energies
3. **Uncertainty Quantification**: Estimating uncertainties in evaluations
4. **Data Comparison**: Comparing experimental vs. evaluated data
5. **Pattern Recognition**: Finding systematic errors or trends

## Quick Example: Visualizing Cross-Section Behavior

Let's create a simple visualization to understand what cross-section data looks like.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create sample energy range (eV)
energy = np.logspace(-2, 7, 1000)  # From 0.01 eV to 10 MeV

# Simulate different cross-section behaviors
# 1/v behavior at low energies (typical for capture)
sigma_capture = 10 / np.sqrt(energy) 

# Resonance structure (simplified)
sigma_resonance = 5 + 100 * np.exp(-((np.log10(energy) - 2)**2) / 0.5)

# Elastic scattering (roughly constant)
sigma_elastic = 5 + 2 * np.sin(np.log10(energy))

# Plot
plt.figure(figsize=(12, 6))
plt.loglog(energy, sigma_capture, label='Capture (1/v)', linewidth=2)
plt.loglog(energy, sigma_resonance, label='Resonance Region', linewidth=2)
plt.loglog(energy, sigma_elastic, label='Elastic Scattering', linewidth=2)

plt.xlabel('Neutron Energy (eV)', fontsize=12)
plt.ylabel('Cross-section (barns)', fontsize=12)
plt.title('Typical Neutron Cross-Section Behavior', fontsize=14)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.show()
```
